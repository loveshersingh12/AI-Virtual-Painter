### Creating random tensors

Random tensors are tensors of some arbitrary size which contain random numbers

```python
# Create two random (but the same) tensors
random_1 = tf.random.Generator.from_seed(7) # set seed for reproducibility
random_1 = random_1.normal(shape=(3, 2))
random_2 = tf.random.Generator.from_seed(7)
random_2 = random_2.normal(shape=(3, 2))


# Are they equal?
random_1, random_2, random_1 == random_2
```

### Getting information from tensors

When dealing with tensors you probably want to be aware of the following attributes:
* Shape
* Rank
* Axis or dimension
* Size

```python
# Create a rank 4 tensor (4 dimensions)
rank_4_tensor = tf.zeros(shape=[2,3,4,5])
rank_4_tensor
```

```python
rank_4_tensor.shape, rank_4_tensor.ndim, tf.size(rank_4_tensor)
```

```python
# Get various attributes of our tensor
print("Datatype of every element:", rank_4_tensor.dtype)
print("Number of dimenisons (rank):", rank_4_tensor.ndim)
print("Shape of tensor:", rank_4_tensor.shape)
print("Elements along the 0 axis:", rank_4_tensor.shape[0])
print("Elements along the last axis:", rank_4_tensor.shape[-1])
print("Total number of elements in our tensor:", tf.size(rank_4_tensor).numpy())
```

### Indexing tensors

Tensors can be indexed just like Python lists.

```python
# Get the first 2 elements of each dimension
rank_4_tensor[:2, :2, :2, :2]
```

```python
# Get the first element from each dimension from each index except for the final one
rank_4_tensor[:1, :1, :4, :1]
```

```python
# Create a rank 2 tensor (2 dimensions)
rank_2_tensor = tf.constant([[10, 7],
                           [3, 4]])

rank_2_tensor.shape, rank_2_tensor.ndim
```

```python
# Get the last item of each row of our rank 2 tensor
rank_2_tensor[:, -1]
```

```python
# Add in extra dimenion to our rank 2 tensor
rank_3_tensor = rank_2_tensor[..., tf.newaxis]
rank_3_tensor
```

```python
# Alternative to tf.newaxis
tf.expand_dims(rank_2_tensor, axis = -1) # -1 means expand the final axis
```

```python
# Expand the 0-axis
tf.expand_dims(rank_2_tensor, axis = 0)
```

** Matrix Multipllication**

In machine learning, matrix multiplication is one of the most common tensor operations.

```python
# Matrix multiplication in tensorflow
print(tensor)
tf.matmul(tensor, tensor)
```

```python
# Matrix multiplication with Python operator "@"
A = tf.constant([[1,2,5],
                 [7,2,1],
                 [3,3,3]])
B = tf.constant([[3,5],
                 [6,7],
                  [1,8]])
```

```python
tf.matmul(A, B)
```

```python
X = tf.constant([[1, 2],
                 [3, 4],
                 [5, 6]])

Y = tf.constant([[7,8],
                 [9, 10],
                 [11, 12]])
X, Y
```

```python
# Let's change the shape of Y
tf.reshape(Y, shape=(2, 3))
```

```python
X.shape, tf.reshape(Y, shape=(2, 3)).shape
```

```python
# Try the multiply X by reshaped Y
X @ tf.reshape(Y, shape=(2, 3))
```

```python
tf.matmul(tf.reshape(X, shape=(2, 3)), Y)
```

```python
# Can do the same with transpose
X, tf.transpose(X), tf.reshape(X, shape=(2, 3))
```

```python
tf.tensordot(tf.transpose(X), Y, axes = 1)
```

```python
# Perform matrix multiplicatio between X and Y
tf.matmul(X, tf.transpose(Y))
```

```python
tf.matmul(X, tf.reshape(Y, shape=(2,3)))
```

```python
tf.tensordot(X, tf.transpose)
```

```python
# Check the values of Y, reshape Y and transposed Y
print("Normal Y:")
print(Y, "\n")

print("Y reshaped to (2,3):")
print(tf.reshape(Y, (2,3)), "\n")

print("Y transposed:")
print(tf.transpose(Y))
```

## Changing the datatype of a tensor

```python
# Create a new tensor with default datatyype (float32)
B = tf.constant([1.7, 7.4])
B, B.dtype
```

```python
c = tf.constant([7, 19])
c, c.dtype
```

```python
# Change from float32 to float16 (reduce precision)
D = tf.cast(B, dtype = tf.float16)
D, D.dtype
```

```python
# Change from int32 to float32
E = tf.cast(c, dtype = tf.float32)
E, E.dtype
```

```python
E_float16 = tf.cast(E, dtype= tf.float16)
E_float16, E_float16.dtype
```

### Aggragating tensors

Aggregating tensors = condensing them from multiple values down to a smaller amount of values

```python
# Get the absolute values
D = tf.constant([-7, -10] )
D
```

# Let's go through the follwoing forms of aggregation:


* Get the minimum
* Get the maximum
* Get the mean of a tensor
* Get the sum of a tensor

```python
tf.size(E), E.shape, E.ndim
```

```python
E
```

```python
# Variance and standard deviation

tf.math.reduce_variance(tf.cast(E, dtype=tf.float32)), tf.math.reduce_std(tf.cast(E, dtype=tf.float32))
```
